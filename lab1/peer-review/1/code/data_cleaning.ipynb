```python
print(raw_data.shape)
df = raw_data
df = df[list(columns_to_keep.keys())]  # only keep relevant columns
df.rename(columns=columns_to_keep, inplace=True)  # rename columns


df = df[df["gcs_total"] >= 14]  # only keep rows with GCS score 14 and 15
df["palpable_skull_fracture"] = df["palpable_skull_fracture"].replace({2: 1})  # combine unclear and palpable skull fracture into one value as that is what the domain knowledge experts did

df.shape
```
